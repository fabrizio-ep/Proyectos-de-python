# Confianza en las instituciones públicas
## 1. Introducción:
La confianza en las instituciones públicas es un elemento crucial para el funcionamiento de la democracia en un país, puesto que influye en la participación ciudadana y la percepción de la población sobre la capacidad del Estado para responder a sus necesidades básicas. Cuando la ciudadanía presenta bajos niveles de confianza en estas instituciones, no solo se dificulta fortalecer la relación entre el Estado y la sociedad, sino que es un síntoma o indicador de que las instituciones son deficientes en cuanto a garantizar protección y servicios a la ciudadanía.
En el Perú, esta problemática resulta relevante debido a las marcadas diferencias económicas, sociales y territoriales que existen entre sus regiones. Factores como la pobreza, las desigualdades en el acceso a servicios públicos, las solicitudes de coimas o sobornos para ser atendidos por las autoridades, corrupción, entre otras problemáticas que son parte del cotidiano de los peruanos, podrían estar fuertemente relacionados con la valoración que tienen las personas sobre las instituciones públicas y democráticas, en específico, en la confianza que reportan hacia ellas. Las implicancias de una baja confianza institucional van desde la impunidad hasta la crisis de representación que se ha manifestado en el Perú en los últimos años, mediante multitudinarias protestas a lo largo y ancho del país. Estas relaciones pueden variar entre departamentos y a lo largo del tiempo, por lo que es importante analizarlas considerando las particularidades regionales y temporales.
En este contexto, la presente investigación busca identificar qué factores están asociados con los niveles de baja confianza en las instituciones públicas en los departamentos del Perú durante el periodo 2021-2025, utilizando información de la Encuesta Nacional de Hogares (ENAHO), específicmanete de los módulos Características de la Vivienda y el Hogar, y Gobernabilidad, Democracia y Transparencia. Para ello, se construirá un indicador de baja confianza y se analizará su relación con distintas variables explicativas mediante estadísticas descriptivas, gráficos y modelos de regresión. Se han elegido 3 variables explicativas: acceso a servicios básicos (agua y electricidad), solicitud de coima en el último año, y percepción de cómo vive el hogar con sus ingresos. Naturalmente surge la pregunta acerca de ¿Por qué consideramos que estas variables podrían estar asociadas con el problema? En cuanto a la primera variable, el acceso a servicios básicos como el agua y la electricidad puede estar relacionado con la confianza institucional, ya que refleja, en cierta medida al ser una proxy, la capacidad del Estado para garantizar condiciones básicas de bienestar a la población. Los hogares que no cuentan con estos servicios podrían percibir una menor presencia o capacidad de respuesta de las instituciones públicas, lo cual podría asociarse con menores niveles de confianza. Respecto a la segunda variable, experiencias como la solicitud de coimas o sobornos durante el último año  pueden afectar directamente la percepción de integridad de las autoridades. Las personas a quienes se les ha solicitado un soborno podrían desarrollar una mayor desconfianza hacia las instituciones públicas, al comprobar que existen prácticas irregulares en la prestación de servicios. En relación a la tercera variable, la situación económica percibida permite aproximarse a su nivel de bienestar económico y a las dificultades que enfrentan para cubrir sus gastos cotidianos. Esta percepción podría influir en la confianza institucional, pues las personas que experimentan mayores dificultades económicas podrían mostrarse más críticas respecto al desempeño de las instituciones públicas. Los hogares que consideran que sus ingresos son insuficientes para cubrir sus gastos podrían mostrar menores niveles de confianza en las instituciones, especialmente si perciben que estas no contribuyen a mejorar su acceso a oportunidades para cambiar su estilo de vida. 
Finalmente, se evaluará la capacidad de un modelo predictivo para estimar el indicador correspondiente al año 2025, comparando sus predicciones con los valores observados.

### 3.1 Construcción de variables (variable dependiente): Indicador de baja confianza
Construimos el indicador de baja confianza en las instituciones públicas. Dado que las respuestas a las preguntas de confianza van desde el 1 (Nada) al 4 (Bastante), definimos baja confianza como respuestas 1 (Nada) y 2 (Poco). Excluiremos las respuestas 5 (No sabe) y 9 (Missing value). Incluiremos las 21 instituciones que aparecen en el cuestionario.
Se clasificó a una persona con baja confianza institucional cuando más del 50 % de sus respuestas válidas correspondía a las categorías Nada o Poco. Para evitar clasificaciones basadas en un número reducido de respuestas, establecimos un mínimo de 11 respuestas válidas de las 21 instituciones. Dicho umbral representa casi la mitad de las instituciones y no exige que la persona haya proporcionado respuestas válidas para las 21 instituciones, puesto que hay algunas que pueden ser desconocidas para la población, por ejemplo la Comisión de Alto Nivel Anticorrupción o la Procuraduría Anticorrupción.
De este modo, se busca mantener un equilibrio entre la consistencia del indicador y la cantidad de observaciones disponibles para el análisis.

```python
base_final = pd.read_csv("base_final.csv", dtype=str, encoding="utf-8-sig")
```

```python
import numpy as np

cols_inst21 = [f"P1${i:02d}" for i in range(1, 22)]
```

```python
respuestas = base_final[cols_inst21].apply(
    pd.to_numeric, errors="coerce"
)

validas = respuestas.isin([1, 2, 3, 4])

base_final["n_validas"] = validas.sum(axis=1)
base_final["n_baja"] = respuestas.isin([1, 2]).sum(axis=1)
```

```python
base_final["prop_baja"] = (
    base_final["n_baja"] /
    base_final["n_validas"].replace(0, np.nan)
).where(base_final["n_validas"] >= 11)


# 1 = baja confianza en más del 50 % de las instituciones
# 0 = no supera ese umbral

base_final["baja_confianza"] = np.where(
    base_final["prop_baja"].isna(),
    np.nan,
    (base_final["prop_baja"] > 0.5).astype(float)
)

# FACTOR DE EXPANSIÓN
base_final["FAMIEGOB07"] = pd.to_numeric(
    base_final["FAMIEGOB07"], errors="coerce"
)

# CÁLCULO DEL PORCENTAJE PONDERADO
def pct_pond(df, var, fac):
    v = df.dropna(subset=[var, fac])
    v = v[v[fac] > 0]

    if v.empty or v[fac].sum() == 0:
        return np.nan

    return (v[var] * v[fac]).sum() / v[fac].sum() * 100

# AGRUPAMOS POR DEPARTAMENTO Y AÑO
conf = (
    base_final
    .groupby(["DEPARTAMENTO", "ANIO"])
    .apply(
        lambda d: pct_pond(
            d, "baja_confianza", "FAMIEGOB07"
        )
    )
    .reset_index(name="pct_baja_confianza")
)
```

El indicador mide el porcentaje de personas que manifiestan baja confianza en las instituciones, al responder “Nada” o “Poco” en más del 50 % de sus respuestas válidas. Los resultados muestran un promedio de 89,82 % en 125 combinaciones de departamento y año.  Los valores oscilan entre 79,49 % y 98,45 %, con una mediana de 90,17 %, lo que evidencia una elevada proporción de baja confianza institucional en las observaciones analizadas

```python
# -- INDICADOR PRINCIPAL: BAJA CONFIANZA -- #
cols_conf = base_final.columns[base_final.columns.str.startswith("P1$")]

base_final["n_validas"] = base_final[cols_conf].isin([1, 2, 3, 4]).sum(axis=1)
base_final["n_baja"] = base_final[cols_conf].isin([1, 2]).sum(axis=1)
base_final["prop_baja"] = base_final["n_baja"] / base_final["n_validas"]

base_final["baja_confianza"] = np.select(
    [base_final["n_validas"] < 11, base_final["prop_baja"] > 0.5],
    [np.nan, 1],
    default=0)

valida = base_final[base_final["baja_confianza"].notna()].copy()
valida["baja_x_factor"] = valida["baja_confianza"] * valida["FAMIEGOB07"]

conf = valida.groupby(["DEPARTAMENTO", "ANIO"])[["baja_x_factor", "FAMIEGOB07"]].sum().reset_index()
conf["pct_baja_confianza"] = conf["baja_x_factor"] / conf["FAMIEGOB07"] * 100
conf = conf[["DEPARTAMENTO", "ANIO", "pct_baja_confianza"]]
```

```python
# -- COIMA -- #
base_final["coima_si"] = base_final["coima"].map({1: 1, 2: 0})

v = base_final[base_final["coima_si"].notna()].copy()
v["coima_x_factor"] = v["coima_si"] * v["fac_parte2"]

coima = v.groupby(["DEPARTAMENTO", "ANIO"])[["coima_x_factor", "fac_parte2"]].sum().reset_index()
coima["pct_coima"] = coima["coima_x_factor"] / coima["fac_parte2"] * 100
coima = coima[["DEPARTAMENTO", "ANIO", "pct_coima"]]

# -- PERCEPCION ECONOMICA: HOGAR EN PEOR SITUACION -- #
base_final["peor_situacion"] = np.select(
    [base_final["percep_cambio_hogar"] == 3, base_final["percep_cambio_hogar"].isin([1, 2])],
    [1, 0],
    default=np.nan)

v = base_final[base_final["peor_situacion"].notna()].copy()
v["peor_x_factor"] = v["peor_situacion"] * v["fac_parte2"]

peor = v.groupby(["DEPARTAMENTO", "ANIO"])[["peor_x_factor", "fac_parte2"]].sum().reset_index()
peor["pct_peor"] = peor["peor_x_factor"] / peor["fac_parte2"] * 100
peor = peor[["DEPARTAMENTO", "ANIO", "pct_peor"]]

# -- ACCESO A SERVICIOS (LUZ + AGUA DE RED, DE 0 A 2) -- #
m100["acc_servicios"] = m100["luz"] + m100["agua_red"]

v = m100[m100["acc_servicios"].notna()].copy()
v["serv_x_factor"] = v["acc_servicios"] * v["FACTOR07"]

servicios = v.groupby(["DEPARTAMENTO", "ANIO"])[["serv_x_factor", "FACTOR07"]].sum().reset_index()
servicios["acc_servicios"] = servicios["serv_x_factor"] / servicios["FACTOR07"]
servicios = servicios[["DEPARTAMENTO", "ANIO", "acc_servicios"]]
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score


encabezados = pd.read_csv("base_final.csv", encoding="utf-8-sig", nrows=0).columns
cols_conf = encabezados[encabezados.str.startswith("P1$")]

columnas = ["DEPARTAMENTO", "ANIO", "FAMIEGOB07", "fac_parte2",
            "coima", "percep_cambio_hogar"] + list(cols_conf)

base_final = pd.read_csv("base_final.csv", encoding="utf-8-sig", usecols=columnas)
m100 = pd.read_csv("m100_limpia.csv", encoding="utf-8-sig",
                   usecols=["DEPARTAMENTO", "ANIO", "luz", "agua_red", "FACTOR07"])

print("Observaciones de ENAHO:", len(base_final))
print("Departamentos:", base_final["DEPARTAMENTO"].nunique())
print("Años:", sorted(base_final["ANIO"].unique()))
```

```python
# Indicador principal: baja confianza institucional
base_final["n_validas"] = base_final[cols_conf].isin([1, 2, 3, 4]).sum(axis=1)
base_final["n_baja"] = base_final[cols_conf].isin([1, 2]).sum(axis=1)
base_final["prop_baja"] = base_final["n_baja"] / base_final["n_validas"]
base_final["baja_confianza"] = np.select(
    [base_final["n_validas"] < 11, base_final["prop_baja"] > 0.5],
    [np.nan, 1], default=0
)

valida = base_final[base_final["baja_confianza"].notna()].copy()
valida["baja_x_factor"] = valida["baja_confianza"] * valida["FAMIEGOB07"]
conf = valida.groupby(["DEPARTAMENTO", "ANIO"])[["baja_x_factor", "FAMIEGOB07"]].sum().reset_index()
conf["pct_baja_confianza"] = conf["baja_x_factor"] / conf["FAMIEGOB07"] * 100
conf = conf[["DEPARTAMENTO", "ANIO", "pct_baja_confianza"]]

# Indicador secundario 1: reporte de corrupción
base_final["coima_si"] = base_final["coima"].map({1: 1, 2: 0})
valida = base_final[base_final["coima_si"].notna()].copy()
valida["coima_x_factor"] = valida["coima_si"] * valida["fac_parte2"]
coima = valida.groupby(["DEPARTAMENTO", "ANIO"])[["coima_x_factor", "fac_parte2"]].sum().reset_index()
coima["pct_coima"] = coima["coima_x_factor"] / coima["fac_parte2"] * 100
coima = coima[["DEPARTAMENTO", "ANIO", "pct_coima"]]

# Indicador secundario 2: empeoramiento percibido del nivel de vida
base_final["peor_situacion"] = base_final["percep_cambio_hogar"].map({1: 0, 2: 0, 3: 1})
valida = base_final[base_final["peor_situacion"].notna()].copy()
valida["peor_x_factor"] = valida["peor_situacion"] * valida["fac_parte2"]
peor = valida.groupby(["DEPARTAMENTO", "ANIO"])[["peor_x_factor", "fac_parte2"]].sum().reset_index()
peor["pct_peor"] = peor["peor_x_factor"] / peor["fac_parte2"] * 100
peor = peor[["DEPARTAMENTO", "ANIO", "pct_peor"]]

# Variable explicativa: acceso a servicios
m100["acc_servicios"] = m100["luz"] + m100["agua_red"]
valida = m100[m100["acc_servicios"].notna()].copy()
valida["serv_x_factor"] = valida["acc_servicios"] * valida["FACTOR07"]
servicios = valida.groupby(["DEPARTAMENTO", "ANIO"])[["serv_x_factor", "FACTOR07"]].sum().reset_index()
servicios["acc_servicios"] = servicios["serv_x_factor"] / servicios["FACTOR07"]
servicios = servicios[["DEPARTAMENTO", "ANIO", "acc_servicios"]]

# Base de análisis a nivel departamento-año
claves = ["DEPARTAMENTO", "ANIO"]
df_regional = pd.merge(conf, coima, on=claves, how="inner")
df_regional = pd.merge(df_regional, peor, on=claves, how="inner")
df_regional = pd.merge(df_regional, servicios, on=claves, how="inner")

datos_2025 = df_regional[df_regional["ANIO"] == 2025].copy()

df_regional.head()
```

```python
X = df_regional[["acc_servicios", "pct_coima", "pct_peor"]]
y = df_regional["pct_baja_confianza"]

X = sm.add_constant(X)

modelo = sm.OLS(y, X).fit()

print(modelo.summary())
```

- pct_coima: Su coeficiente es de 0.25, de signo positivo como se esperaba; por cada punto porcentual adicional de personas a las que les pidieron coima, la baja confianza sube en promedio 0.25 puntos porcentuales. Pero no es significativo (p = 0.290).
- pct_peor: Su coeficiente es de −0.15, de signo negativo, el único significativo (p < 0.001). Si el porcentaje de hogares que dice estar peor sube 10 puntos, la baja confianza baja en promedio 1.5 puntos, lo contrario a lo esperado.

- H3: el signo negativo de pct_peor podría deberse a que las dos variables tienen tendencias opuestas en el tiempo (la percepción económica mejora mientras la confianza empeora) y no a una relación directa entre ellas.

### Hipótesis:
- H1: la baja confianza en las instituciones responde más a factores territoriales persistentes (historia, presencia estatal, conflictividad) que a las variables medidas aquí, ya que el orden entre departamentos se mantiene entre años.
- H2: la experiencia directa de corrupción podría estar asociada a mayor desconfianza (el signo es positivo), pero la coima es poco frecuente (alrededor de 2.4 % de personas) y la relación no es estadísticamente significativa a nivel departamental.
- H3: el aumento de 2025 parece ser un fenómeno general, con factores nacionales comunes a todos los departamentos que las variables regionales no captan. Habría que explorar variables de contexto político o de coyuntura.
- H4: el comportamiento de la percepción económica (que mejora mientras la confianza empeora) sugiere que la desconfianza no parece explicarse por la percepción de la situación económica del hogar.

### Limitaciones:
- Solo hay 25 observaciones de prueba y 100 de entrenamiento, por lo que las métricas son inestables.
- Las muestras departamentales son pequeñas en algunos casos (por ejemplo, Madre de Dios, con unos 600 registros por año), lo que genera ruido en los porcentajes.
- Los resultados son asociaciones, no efectos causales.
